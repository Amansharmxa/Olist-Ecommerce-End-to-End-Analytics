```python
import os
import pandas as pd

RAW_DIR = "rawdata_csv/"

# 1. Dekhte hain folder me total kitni aur kaun-kaun si files hain
files = os.listdir(RAW_DIR)
print("--- Available Files in Folder ---")
for f in files:
    if f.endswith('.csv'):
        print("-", f)

print("\n" + "="*40 + "\n")

# 2. Key files ko load karke unka shape aur missing values check karte hain
file_names = {
    "orders": [f for f in files if 'orders_dataset' in f][0],
    "order_items": [f for f in files if 'order_items_dataset' in f][0],
    "customers": [f for f in files if 'customers_dataset' in f][0],
    "products": [f for f in files if 'products_dataset' in f][0],
    "payments": [f for f in files if 'order_payments_dataset' in f][0]
}

dfs = {}
for key, filename in file_names.items():
    path = os.path.join(RAW_DIR, filename)
    dfs[key] = pd.read_csv(path)
    print(f"File: {key} ({filename})")
    print(f"  - Shape (Rows, Columns): {dfs[key].shape}")
    print(f"  - Missing Values: {dfs[key].isnull().sum().sum()}")
    print("-" * 30)
```

```python
import os
import pandas as pd

RAW_DIR = "rawdata_csv/"
PROCESSED_DIR = "processed_data/"
os.makedirs(PROCESSED_DIR, exist_ok=True)

files = os.listdir(RAW_DIR)
print("--- Saari 9 Files ki Cleaning Shuru ho Rahi hai ---")

for f in files:
    if f.endswith('.csv'):
        file_path = os.path.join(RAW_DIR, f)
        df = pd.read_csv(file_path)
        
        # 1. Basic cleaning: Remove exact duplicate rows
        df = df.drop_duplicates()
        
        # 2. Specific file ke hisab se date columns convert karna
        date_columns = [col for col in df.columns if 'timestamp' in col or '_at' in col or 'date' in col]
        for col in date_columns:
            try:
                df[col] = pd.to_datetime(df[col], errors='coerce')
            except:
                pass
                
        # 3. Cleaned data ko processed folder me save karna
        output_path = os.path.join(PROCESSED_DIR, f"clean_{f}")
        df.to_csv(output_path, index=False)
        print(f"[SUCCESS] Cleaned & Saved: clean_{f} (Rows: {df.shape[0]}, Cols: {df.shape[1]})")

print("\n--- Sabhi 9 files successfully clean ho kar 'processed_data' folder me save ho gayi hain! ---")
```
