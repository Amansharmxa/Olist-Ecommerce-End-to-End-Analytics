```python
from sqlalchemy import create_engine
import pandas as pd
import os

# 1. Database Connection String 
# (Note: 'postgres' ki jagah apna username aur 'your_password' ki jagah apna pgAdmin password likhein)
DB_USER = "postgres"
DB_PASSWORD = "aman123"  # Yahan apna password daalein
DB_HOST = "localhost"
DB_PORT = "5432"
DB_NAME = "olist_ecommerce"

# Connection create kar rahe hain
engine = create_engine(f'postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}')

# 2. Processed data folder ka path
PROCESSED_DIR = "processed_data/"

# 3. Loop chalakar saari CSV files ko PostgreSQL tables me convert karna
print("--- PostgreSQL Database Import Shuru ho raha hai ---\n")

for file in os.listdir(PROCESSED_DIR):
    if file.endswith('.csv'):
        # File name se table name banana (jaise clean_olist_customers_dataset.csv -> olist_customers_dataset)
        table_name = file.replace('clean_', '').replace('.csv', '')
        
        file_path = os.path.join(PROCESSED_DIR, file)
        df = pd.read_csv(file_path)
        
        # DataFrame ko PostgreSQL me table ke roop me push karna
        df.to_sql(table_name, engine, if_exists='replace', index=False)
        print(f"[SUCCESS] Table created in Database: '{table_name}' ({df.shape[0]} rows loaded)")

print("\n--- Sabhi 9 tables successfully PostgreSQL me load ho gayi hain! ---")
```
